```python
# importing libraries
import numpy as np 
from PIL import Image
```

```python
img = r"C:\Users\Surface Pro 7\OneDrive\Desktop\Python-logo (1).png"
```

```python
img
```

```python
img = Image.open(img)
```

```python
img
```

```python
img.size
```

```python
img = img.resize((255,255))
```

```python
img
```

```python
i_a = np.array(img)
```

```python
i_a
```

```python
i_c = img.convert('L')
```

```python
i_c
```

```python
i_a1 = np.array(i_c)
```

```python
i_a1
```

```python
from sklearn.decomposition import PCA
```

```python
p = PCA(n_components=58)
```

```python
p
```

```python
p1 = p.fit_transform(i_a1)
```

```python
p1
```

```python
import matplotlib as mpl

mpl.rcParams['font.enable_last_resort'] = False
```

```python
plt.imshow(p1)
plt.show()
```

```python
p2 = p.inverse_transform(p1)
```

```python
p2
```

```python
plt.imshow(p2)
plt.show()
```
